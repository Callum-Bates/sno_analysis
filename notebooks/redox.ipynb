
# Redox PTM predictors recycle stale data

**At least 75 distinct predictors for redox-type PTMs can be named (verified 46, unverified 29), but the field is a crowded, ageing sequence-window-classifier literature built on a few small, reused datasets, and it is thin exactly where a design-oriented PhD would need it.** The count is a lower bound: five parallel passes and two follow-ups were hampered by blocked PMC, PubMed, Europe PMC and ScienceDirect pages, so many entries were seen only as titles. S-nitrosylation (at least 17 tools) and S-sulfenylation (at least 19) are crowded, and their headline metrics are not comparable across papers. S-glutathionylation has had no new dedicated predictor since 2020 (as far as limited searches show), S-sulfinylation has one, persulfidation has two, methionine oxidation about two or three, carbamylation one (2013), and sulfonylation, succination and 4-HNE adducts none that I could find. No independent benchmark of redox PTM predictors was found anywhere, and new-tool output looks like it fell from roughly 6 dated tools per year (2016-2021) to roughly 2 per year (2022-2026), though recall differs between those periods. Generalist PTM language models mostly ignore redox chemistry: of those checked, only MeToken, CLASPP and LMPTMSite include any redox class. For a PhD on enzymes that sense their chemical environment through nPTMs, the answer is yes with a caveat. The field is a useful source of data, baselines and a defensible benchmarking or data-curation contribution, but adding another sequence-window classifier to S-nitrosylation or sulfenylation would be hard to defend. The design side of the literature was barely searched, so no claim about a gap there is supported. A verify-first list closes the report.

## How to read the counts and ratings

Counts are written "at least N (verified V, unverified U)". Verified means a source the researchers actually opened named the tool and gave at least a year or method; unverified means the tool was seen only as a title, search hit or bare name. Evidence quality is rated by how much of the row rests on read primary or near-primary text: Medium means several tools were read with data sizes and metrics, Low means mostly titles or a negative result from one or two searches. "None found" claims are weaker than they sound, because the searches behind them were few, were run through a general web search engine, and could not query Europe PMC, GitHub search or the preprint servers directly. Multi-PTM tools are counted once: pCysMod is counted under S-nitrosylation and DeepNitro under S-nitrosylation (it also covers tyrosine and tryptophan nitration).

## Two cysteine modifications are crowded, and the crowding is mostly the same data

**S-nitrosylation** has the longest history: SNOSID (2006), GPS-SNO (2010), SNOSite (2011) and iSNO-PseAAC (2013), then DeepNitro (2018) and PreSNO (2019) ([pLMSNOSite intro](https://bmcbioinformatics.biomedcentral.com/articles/10.1186/s12859-023-05164-9)). Those were followed by RecSNO, Mul-SNO, RF-SNOPS and pCysMod around 2020-2021, pLMSNOSite in 2023, SNO-DCA in January 2024 and the NitroXAI software release in 2026. DeepNitro trained on **3,409 positive and 17,453 negative cysteines** with a 485/4,947 independent test, curated from literature before 30 June 2015, and its own cross-validation AUC for SNO was only **0.70** ([DeepNitro](https://academic.oup.com/gpb/article/16/4/294/7225049)). Reuse of this lineage is verified for two models: pLMSNOSite explicitly adopts the DeepNitro data "via PreSNO" ([pLMSNOSite](https://bmcbioinformatics.biomedcentral.com/articles/10.1186/s12859-023-05164-9)), and pCysMod drew on sources including GPS-SNO, DeepNitro and SNOSite ([pCysMod](https://www.frontiersin.org/articles/10.3389/fcell.2021.617366/full)). Reuse by RecSNO, Mul-SNO, SNO-DCA and RF-SNOPS was not checked, and the 2025 paper that reportedly reuses RecSNO and dbSNO v2.0 data (S0003269725001915) is known only from your brief and was never opened. "The DeepNitro dataset" is also not one fixed file: DeepNitro reports 3,409 positives, while the pLMSNOSite re-split after cd-hit at 0.3 identity has 3,734 positives and 20,333 negatives.

The best-documented recent SNO model shows how modest the ceiling is. pLMSNOSite (ProtT5 embeddings stacked with a word-embedding 2D-CNN) reports an independent-test MCC of **0.340**, sensitivity 0.735 and specificity 0.773 on 351 positives and 3,168 negatives. At that class ratio those figures imply a precision of only about 26% (my arithmetic: about 258 true positives against about 719 false positives). The 0.754 that the fetched text labels as AUROC equals the geometric mean of sensitivity and specificity (0.754), so it is almost certainly G-mean, not AUROC; treat the AUROC as unreported ([pLMSNOSite](https://par.nsf.gov/servlets/purl/10510992)). NitroXAI (ESM-2 residue embeddings fused with handcrafted features) has a Zenodo release dated 2 September 2026 but no paper, no dataset sizes and no performance numbers, and its claim that the handcrafted-only SNO-CLIM variant matches PLM models with about 1,000 times fewer parameters is unquantified ([GitHub](https://github.com/notfromworld/NitroXAI), [Zenodo](https://zenodo.org/records/22238589)).

**S-sulfenylation** is crowded to a similar degree. DeepCSO's comparison list gives ten earlier tools: MDD-SOH, iSulf-Cys, SOHSite, PRESS and SOHPRED (all 2016), S-SulfPred and SulCysSite (2017), Sulf_FSVM and PredCSO (2018) and Fu-SulfPred (2019) ([DeepCSO](https://www.ncbi.nlm.nih.gov/pmc/articles/PMC7736615/)). To these can be added SIMLIN (2019; 1,235 positives and 9,349 negatives in human, AUC 0.82), DBN-Sulf (2018; 1,007/7,837, AUC 0.80), DeepCSO (2020; human 2,758 positives, Arabidopsis AUC 0.852) and, seen only as titles or snippets, SVM-SulfoSite, SulSite-GTB, fastSulf-DNN, nSSPred and BiGRUD-SA ([SIMLIN](https://bmcbioinformatics.biomedcentral.com/articles/10.1186/s12859-019-3178-6), [DBN-Sulf](https://benthamdirect.com/content/journals/cbio/10.2174/1574893612666171122152208)). The newest, BioSemAF-BiLSTM (Frontiers in Genetics, 2025), still trains and tests on the iSulf-Cys benchmark of 1,045 positives and 7,124 negatives. Its headline **auROC of 0.96 is not comparable** with older results: its test set has 145 positives against 268 negatives (about 1:1.8), whereas its training set is about 1:7.6 ([Frontiers in Genetics](https://www.frontiersin.org/journals/genetics/articles/10.3389/fgene.2025.1616880/full)). Two notes disagree on the DeepCSO Arabidopsis positive count (1,380 in the paper text read, 1,535 in another note); the paper figure is the better-sourced one.

The same pattern appears in both: new models re-measure on old data with new encoders (CNN, then RNN, then attention, then embeddings), and every comparison I could find was run by the authors of the new method. The 2025 plant redox review says the same thing directly: "standardized benchmarking across datasets is lacking" and transient modifications are prone to false positives ([PMC12295296](https://www.ncbi.nlm.nih.gov/pmc/articles/PMC12295296/)).

## The open modifications are open mostly because the data are small

**Glutathionylation** has at least five dedicated tools: Sun et al. 2013 (75 positive and 158 negative cysteines, AUC 0.879), PGluS 2015 (accuracy about 71%), GSHSite 2015 (1,783 positive and 8,423 negative training sites from one mouse macrophage study; independent MCC 0.56 against cross-validation MCC 0.32), Anashkina 2020 (221 positives; specificity 40%, MCC 0.133, no server) and DeepGSH (about 2020, year unverified, servers listed in CysModDB) ([Sun 2013](https://www.ncbi.nlm.nih.gov/pmc/articles/PMC3572087/), [GSHSite](https://www.ncbi.nlm.nih.gov/pmc/articles/PMC4388702/), [PGluS](https://pubs.rsc.org/en/content/articlelanding/2015/mb/c4mb00680a), [Anashkina](https://pmc.ncbi.nlm.nih.gov/articles/PMC7489215)). No dedicated 2021-2026 glutathionylation predictor turned up in several differently worded searches, so the subfield looks stale. It also looks weak on sequence signal alone: Anashkina's MCC of 0.133 is the only independent-looking evidence, and GSHSite's independent score exceeding its cross-validation score points to an easier or leaky test set (my reading, not a published claim). PGluS's year is 2015 in one note and 2014 in another; the RSC URL used 2015. A separate warning from the notes: iGlu_AdaBoost, DeepDN_iGlu, GlutPred and similar names predict lysine glutarylation, not cysteine glutathionylation, and must not be counted.

**Persulfidation** has two models. pCysMod's S-sulfhydration module (2021; 2,535 sites on 1,707 proteins, AUC 0.793) and Sul-BertGRU (Bioinformatics 41(3), btaf078, February 2025; 19,402 samples with 2,705 positives, sensitivity 85.82%, specificity 68.24%, MCC 0.5513, AUC 0.7703) ([pCysMod](https://www.frontiersin.org/articles/10.3389/fcell.2021.617366/full), [Sul-BertGRU](https://academic.oup.com/bioinformatics/article/41/3/btaf078/8026686)). The 2025 plant review states that for persulfidation "no predictive programs have been developed," which contradicts both entries and shows how unreliable single-source "none exist" statements are in this area ([PMC12295296](https://www.ncbi.nlm.nih.gov/pmc/articles/PMC12295296/)). Experimentally the area is growing: a 2024 meta-analysis counted 16 persulfide proteomics studies from 2010-2023 using 10 different enrichment methods ([Li, Stoltzfus, Michel 2024](https://pubs.rsc.org/en/content/articlehtml/2024/cb/d3cb00106g)). Which of these studies fed either predictor was not established, and method heterogeneity is a plausible source of label noise (inference).

**Sulfinylation and sulfonylation** are effectively one paper. DLF-Sul (2022) trains on a balanced 520/520 subset of 742 sulfinylation sites on 538 proteins, with a balanced 111/111 test set, reporting MCC 0.84 and AUC 0.964 ([DLF-Sul](https://academic.oup.com/bib/article/23/5/bbac323/6658856)). That MCC applies to a balanced 222-site test and says little about proteome-wide use. No sulfonylation predictor was found. The pCysMod note gives 655 sulfinylation sites on 538 proteins for the same dataset where DLF-Sul says 742; the discrepancy is unresolved (see the data section).

**Tyrosine nitration** has at least nine tools from GPS-YNO2 (2011) to NTpred (2023), and two benchmarks recur. The numbers show how split-dependent the field is: PredNTS (2021) reports an independent AUC of 0.860 but MCC of only **0.286**, while NTpred reports an MCC of 0.849 on one split and 0.570 on the other ([PredNTS](https://www.mdpi.com/1422-0067/22/5/2704), [NTpred](https://www.biorxiv.org/content/10.1101/2023.02.18.529069v1.full)). The two sources describe the benchmark train and test sizes inconsistently, and this was left unreconciled. NTpred's own authors also describe a systematic false-positive bias, with specificity far above sensitivity.

**Carbonylation** has at least 11 tools (CarSPred 2014, iCar-PseCp 2016, PTM-ssMP about 2017, iCarPS 2020, CarSite-II 2021, PreCar_Deep 2021, a 2022 deep-network study, Carsite_AGan 2023, an Advanced Science attention model 2025, plus the older CarSite and PTMPred comparators), yet its datasets are minuscule. CarSPred's training positives are K 266, R 119, T 116 and P 114, with AUCs of only **0.69-0.71** despite roughly 85% accuracy ([CarSPred](https://pmc.ncbi.nlm.nih.gov/articles/PMC4210226/)). CarSite-II's independent test has just 16 P and 24 T positives, and its per-residue MCC ranges from 0.29 (P) to 0.64 (K) ([CarSite-II](https://link.springer.com/article/10.1186/s12859-021-04134-3)). CarbonylDB, the standard data source, holds 1,495 proteins and 3,781 sites and was last updated 28 February 2018 ([CarbonylDB](https://carbonyldb.missouri.edu/CarbonylDB/index.php)). One more check item: Carsite_AGan lists 1,175 test positives from "CarSite2 data", which equals the sum of CarSite-II's human training positives (618+162+204+191), while its 16,533 negatives do not match that set's 96,533; it may be reusing training data as a test set, but this needs the paper read ([Carsite_AGan](https://pmc.ncbi.nlm.nih.gov/articles/PMC10644465/)). The Advanced Science paper (PMID 40145846) could not be read at all.

**Methionine oxidation** rests on a 2017 random-forest study from one dataset of 122 positive and 853 negative Met residues in Jurkat cells (accuracy 0.747, sensitivity 0.682), with structural features such as solvent accessibility ranking highest ([BMC Bioinformatics 2017](https://pmc.ncbi.nlm.nih.gov/articles/PMC5622526)). MetOSite (2019) now holds 7,242 sulfoxide sites on 3,562 proteins ([MetOSite](https://academic.oup.com/bioinformatics/article/35/22/4849/5514041)), so a larger model is feasible but untested in anything found. Notes disagree on whether anything followed: one says no successor exists, another surfaced "MetODeep 2019" and antibody Met-oxidation risk models (2018-2021) as titles only. Met oxidation is also a known sample-preparation artefact, so "not oxidised" labels are weak (inference).

## Comparison points show that tool count tracks community interest, not data

Palmitoylation has at least 14 tools but about 3,000 sites: GPS-Palm (2020) uses 3,098 sites on 1,618 proteins from 79 species (AUC 0.855 general), and pCysMod's palmitoylation module has 2,766 sites ([GPS-Palm](https://www.biocuckoo.cn/pub/GPS-Palm.pdf)). That is only slightly more data than persulfidation (2,535 sites) with several times as many tools, so scarcity of tools in the open areas is partly interest and not only data (inference). Glycation (lysine) has at least eight tools from GlyNN (2006) to the 2022 language-model work, and GlyStruct's benchmark is 235 glycated and 1,518 non-glycated lysines; BERT-Kgly pre-trains a BERT on 556,603 UniProt sequences and PTG-PLM tests six protein language models ([GlyStruct](https://link.springer.com/article/10.1186/s12859-018-2547-x), [BERT-Kgly](https://www.ncbi.nlm.nih.gov/pmc/articles/PMC9580886/), [PTG-PLM](https://github.com/Alhasanalkuhlani/PTG-PLM)). Carbamylation has exactly one model: a 2013 one-class kNN for carbamylated lysine (280 features; independent sensitivity 66.67%, specificity 100%, MCC 0.81 on an evidently tiny set) ([Mol BioSyst 2013](https://pubs.rsc.org/en/content/articlelanding/2013/mb/c3mb70195f)). The one hard datum on carbamylation data quality is that its +43.0058 Da mass shift is confusable with citrullination, and antibody reagents often cross-react ([Verheul 2018](https://pure.amsterdamumc.nl/en/publications/pitfalls-in-the-detection-of-citrullination-and-carbamylation-2/)). For succination and 4-HNE adducts, the single search returned only lysine succinylation tools.

## Data resources are overlapping, partly static, and contradict each other

The curated cysteine databases are CysModDB (2022: **70,536** sites on 21,654 proteins, 12 organisms, 12 modification types, downloadable in several formats, with only 17 citations as of October 2026), iCysMod (2021), dbSNO 3.0 and, newest, CysDBase (May 2026) ([CysModDB](https://academic.oup.com/bib/article/23/6/bbac460/6775608), [Database Commons](https://ngdc.cncb.ac.cn/databasecommons/database/id/8301)). They overlap by design (CysModDB explicitly positions itself against iCysMod, SwissPalm and dbSNO), so counts must not be summed and predictors trained on them inherit a common ~2020-2022 literature ceiling (inference). Older single-PTM resources are still in use: dbGSH (2014; 2,257 sites), RedoxDB (2012; 2,203 sites), CarbonylDB (2018) and MetOSite (2019) ([iCysMod/bbaa400 table](https://academic.oup.com/bib/article/22/5/bbaa400/6066620)). Quantitative sources differ in kind from binary site lists: CysDB covers 62,888 human cysteines for reactivity and ligandability across nine studies (a ChemRxiv preprint; peer-reviewed version not confirmed), and OxiMouse gives cysteine oxidation percentages across ten mouse tissues in young and old animals ([CysDB](https://chemrxiv.org/engage/chemrxiv/article-details/63edc76cfcfb27a31fdedfde), [OxiMouse](https://oximouse.hms.harvard.edu/help.html)).

## Benchmarking is the field's weakest point, with specific, checkable problems

No community benchmark, leaderboard or independent head-to-head evaluation of redox PTM predictors was found, and the 2025 review confirms benchmarking is lacking ([PMC12295296](https://www.ncbi.nlm.nih.gov/pmc/articles/PMC12295296/)). The concrete problems, each supported by at least one read source, are as follows. Negatives are almost always unmodified cysteines (or K, R, P, T, Y, M) in the same proteins; because probe-based detection is condition- and abundance-dependent, they are better treated as unlabelled (positive-unlabelled), which is my inference, since no read source quantifies it. Test sets are the authors' own hold-outs from the same database, and balanced undersampling is common (DLF-Sul's 111/111 test; BioSemAF's 1:1.8 test), which inflates MCC and auROC relative to real class ratios. Redundancy thresholds differ widely (40% identity for BioSemAF and DeepNitro, 30% for pLMSNOSite and iCarPS, 90% for DLF-Sul), so a 90% CD-HIT cut leaves substantial redundancy and likely optimistic estimates (inference). Benchmarks recur across papers (iSulf-Cys 2014 for sulfenylation, CarSPred for carbonylation, DeepNitro for SNO), so reported gains are not independent evidence. High accuracy coexists with low AUC or MCC (CarSPred), and independent test sets for rare residues are tiny (5 T and 12 P positives in CarSPred). Finally, performance depends heavily on how a split is made (PredNTS vs NTpred above). An adjacent field shows the failure mode: ligandable-cysteine models trained on crystal structures reached 96% AUC and fell to 0.70-0.75 on chemoproteomic data ([bioRxiv 2023](https://www.biorxiv.org/content/10.1101/2023.08.17.553742.full.pdf)).

Not found in anything read, and so not supportable from these notes: any quantification of label noise in biotin-switch or dimedone-probe data, the structural features (pKa, solvent exposure, acid-base motif) that current SNO predictors use, and any 2022-2026 critical review of SNO predictors. The planned review (PMC7892234) was never opened. pLMSNOSite states that AlphaFold2 structure could improve it, i.e. it is sequence-only ([pLMSNOSite](https://par.nsf.gov/servlets/purl/10510992)).

So the chemistry coverage of generalists is thin and accidental: it follows what dbPTM or UniProt annotations happen to contain, and "oxidation" or "sulfoxidation" classes there probably merge Met and Cys oxidation (inference), which makes them weak for redox-switch questions. No "ProteinGym for PTMs" leaderboard was found. In short, a redox-specific model has little direct competition from generalists; the competition is the older single-PTM tools, and the constraint is label quality.

Two notes disagreed about what counts as a protein language model, and the line matters for any "no PLM-based predictor exists" claim. Sul-BertGRU is now resolved by a direct check of the Bioinformatics paper: its BERT is trained from scratch on the task with a masked-language objective (15% of tokens masked), not pretrained on a large protein corpus, so it is not a PLM in the ESM/ProtT5 sense. The same check confirmed its data (19,402 samples, 2,705 positives, drawn from iCysMod) and metrics (sensitivity 85.82%, specificity 68.24%, MCC 0.5513, AUC 0.7703). BioSemAF-BiLSTM uses fastText embeddings and PSSM and no PLM, though a third note lists it loosely as a "feature extraction framework". BERT-Kgly pre-trains its own BERT on 556,603 UniProt sequences, which is a PLM only in a narrow sense. Only pLMSNOSite (ProtT5) and NitroXAI (ESM-2) verifiably use a pretrained PLM among redox site predictors, and NitroXAI has no paper. No graph-based, AlphaFold-structure-based or LLM-based redox site predictor was found, with the limitation that GitHub and Hugging Face coverage was shallow.

## The trend: a 2016-2021 peak, then a trickle

Counting tools with a known or reasonably inferred year (a lower bound; some years are inferred and the late-period search was more targeted than the early-period list-mining), the redox core shows roughly 6 new tools in 2016, 5 in 2017, 5 in 2018, 4 in 2019, 7 in 2020 (one year unverified) and 7 in 2021, then 3 in 2022, 3 in 2023, 1 in 2024, 3 in 2025 and 1 in 2026 (NitroXAI, software only). That is about 34 tools in 2016-2021 against about 11 in 2022-2026. The verified 2023-2026 outputs are pLMSNOSite, NTpred and Carsite_AGan (2023), SNO-DCA (2024), Sul-BertGRU, BioSemAF-BiLSTM and the Advanced Science carbonylation model (2025), and NitroXAI (2026). Reading the drop as "interest peaked" is reasonable; reading it as "nothing new exists" is not, because a missed paper would not be caught by the search engine used.

## Opportunity: what is defensible for a design-focused PhD

The central point is that site predictors answer a different question from the one your project asks. They are binary, mostly sequence-window classifiers of whether a cysteine is annotated as modified. An enzyme that senses its chemical environment through an nPTM needs to know how reactive a particular cysteine or lysine is in a particular fold, what the modification does to structure and activity, and how to build that into a design. Nothing in the notes shows a site predictor used inside a design pipeline, and the design literature was searched too thinly to claim a gap. The verified design precedents are reporter engineering such as OxyR-cpGFP (oROS-G, a preprint) and computationally designed redox-potential platforms such as the de novo heme maquette ([oROS-G](https://stemjnl.org/preprints/PPR800967), [PNAS 2023](https://par.nsf.gov/servlets/purl/10648407)). Extending to catalytic enzymes gated by an nPTM looks open, but that is a hypothesis. Structure tools are only partly ready: AlphaFold 3 accepts modified residues by CCD code, but no benchmark on cysteine oxidation, nitrosylation, carbamylation or glycation was found, and a 2025 preprint shows AF2, AF3, Boltz-1 and others systematically form wrong disulfides for a Cys-thioether crosslink ([AF3 inputs](https://raw.githubusercontent.com/google-deepmind/alphafold3/main/docs/input.md), [Lacombe 2025](https://arxiv.org/pdf/2503.17368)).

Options ranked by how defensible they are:

| Rank | Option | Why defensible | Main risk |
|---|---|---|---|
| 1 | **Independent, leakage-controlled benchmark of existing redox predictors** (SNO, sulfenylation, persulfidation first): homology-split test sets, real class ratios, PU-aware negatives, MCC and AUPR | Reviewers state that standardised benchmarking is lacking, no independent benchmark was found, every published comparison is author-run, and metrics are demonstrably non-comparable (BioSemAF, DLF-Sul, RF-SNOPS, pLMSNOSite). Needs no new wet data, and suits a period of learning scikit-learn and PyTorch | Many servers may be dead or unmaintained (none were tested), so tools must be rerun from code or reimplemented; "no independent benchmark exists" rests on limited searching |
| 2 | **Refreshed, quality-audited dataset** (SNO first, then glutathionylation, persulfidation, carbonylation) built from post-2015 literature and new proteomics | DeepNitro-lineage positives stop at June 2015, CarbonylDB at 2018, and CysModDB and CysDBase are keyword- or literature-curated with unresolved count conflicts; new redox proteomics keeps appearing (16 persulfidation studies by 2023). It is also the foundation for options 1 and 3 | Label quality: probe-detected sites vs real modification, and the CysDBase and dbSNO conflict must be settled first; curation is slow |
| 3 | **Structure- and microenvironment-aware reactivity or occupancy model** combining AlphaFold structures with CysDB reactivity and OxiMouse occupancy data (predicting how reactive or oxidised a cysteine is, not just whether it is annotated) | No verified redox site predictor uses structure or graph models; pLMSNOSite names AlphaFold2 as an untried improvement; ligandability models already show structure helps (AUC 0.93-0.94 in distribution) and bridges directly to design. Occupancy data support regression tasks few published tools target | Ligandability models collapse on cellular data (0.70-0.75), so out-of-distribution validation is essential; novelty is not confirmed (2023-2026 coverage was shallow, and a 2026 AF3 redox-state modelling paper was seen by title only) |
| 4 | **Unified multi-task redox-cysteine model on a modern backbone**, covering SNO, sulfenylation, glutathionylation, persulfidation and sulfinylation | pCysMod (2021) is the only published multi-cysteine model and omits glutathionylation; generalists lack these classes; multi-task sharing helps small classes | The gap rests on absence searches; the small-class data are noisy; NitroXAI's unquantified claim that PLMs add little warns that gains may be small; MeToken and CLASPP per-class results were not read |
| 5 | **Dedicated persulfidation or glutathionylation predictor** | Few or stale competitors (two and five) | Sequence signal looks weak (Anashkina MCC 0.13; Sul-BertGRU AUC 0.77, specificity 68%), data are small and noisy; a sixth sequence classifier is hard to publish convincingly without option 2 first |
| 6 | **Carbamylation (or other near-empty non-enzymatic PTMs) as a modelling target** | One 2013 model; the most thematically aligned with non-enzymatic PTMs | Almost no public data, MS ambiguity with citrullination; feasible only if your experimental collaborators in Durham can supply labelled data, which the notes did not assess |
| 7 | **Another SNO or sulfenylation sequence-window classifier** | Easy to build | Crowded (17 and 19 tools), data recycled, no independent benchmark to prove improvement; least defensible |

Options 1 and 2 reinforce each other and can be done in sequence, with the curated data feeding the benchmark, and option 3 is the one that connects most directly to enzyme design. That ordering is my judgement from the notes, not a finding from them.

## Conclusion

The picture that emerges is of a field that looks busy by tool count but is shallow in evidence: the same few benchmarks, authors' own tests, and metric choices that make 0.96 auROC and 0.34 MCC describe tasks of similar difficulty. What this means for you is that the opening lies in measurement and data, not in model novelty. A careful benchmark with leakage control and a refreshed dataset would be new, and either would give you honest baselines before you commit to any design strategy. The second takeaway is a boundary on what this report can say. The tool counts rest on a mix of read papers and search hits, the "open" modifications are only as open as one or two searches show, and the design-of-nPTM-sensing-proteins literature, the part closest to your PhD, was not searched adequately and should be treated as unknown, not empty.
